### Comando #1: `cp` → ¡Copia archivos como un duplicador mágico!

`cp` = **Copy** (copiar).

¡Es como sacar una fotocopia perfecta de un tesoro!

Primero creamos un archivo de práctica (si no tienes la carta del módulo anterior)

```python
!echo "Este es mi tesoro principal de El Salvador 🏄‍♂️🌋" > tesoro_principal.txt
!cat tesoro_principal.txt
```

```python
# Copia el tesoro (copia de seguridad)
!cp tesoro_principal.txt copia_de_seguridad.txt
```

```python
# ¡Mira! Ahora tienes dos
!ls
!cat copia_de_seguridad.txt
```

También puedes copiar carpetas completas con `-r` (recursivo):  
`!cp -r Tesoros_ElSalvador Tesoros_Copia_Segura`

### Comando #2: `mv` → ¡Mueve o renombra archivos como un teletransportador!

`mv` = **Move** (mover).

¡Sirve para mover a otra carpeta o simplemente cambiarle el nombre!

```python
# Renombra la copia
!mv copia_de_seguridad.txt backup_importante_2025.txt
```

```python
!ls
```

¡Listo! Cambiaste el nombre sin perder el contenido.

### Comando #3: `rm` → ¡Elimina archivos... con mucho cuidado! ⚠️

`rm` = **Remove** (eliminar).

**¡ADVERTENCIA NINJA!** En Linux NO hay papelera de reciclaje. Lo que borras... ¡desaparece para siempre!

Siempre piensa dos veces y usa `-i` para que te pregunte confirmación.

```python
# Elimina con confirmación (seguro)
!rm -i backup_importante_2025.txt
```

```python
!ls
```

**Nunca uses:** `rm -rf /` (¡eso borraría todo el sistema! 😱)

Regla de oro: **Si no estás 100% seguro, no lo borres.**

Hoy aprendiste poderes ninja:
- `cp` → copiar (hacer backups)
- `mv` → mover o renombrar
- `rm` → eliminar (¡con cuidado!)
