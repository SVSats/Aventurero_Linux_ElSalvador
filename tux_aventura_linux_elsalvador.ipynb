# 🐧 Aventura Linux - Tu perfil de aventurero/a en El Salvador 🇸🇻

¡Bienvenido/a a esta aventura interactiva con Tux, la mascota de Linux!

## Instrucciones

1. Ejecuta esta celda de código (Shift + Enter)
2. Sigue las instrucciones que aparecerán abajo
3. Responde las preguntas cuando te las pidan
4. ¡Disfruta de tu aventura Linux!

````python
# 🐧 AVENTURA LINUX - TU PERFIL DE AVENTURERO/A
print("=" * 80)
print("🐧 ¡BIENVENIDO/A A LA AVENTURA CON TUX EN EL SALVADOR! 🇸🇻".center(80))
print("=" * 80)

# Paso 1: Nombre
while True:
    nombre = input("🐧 Tux pregunta: ¿Cómo te gusta que te llamen, valiente aventurero/a? ").strip()
    if nombre and any(c.isalpha() for c in nombre):
        break
    print("⚠️ Por favor escribe un nombre válido (letras y espacios).")

print(f"\n🎉 ¡Hola {nombre}! Tux está emocionado de conocerte. 🐧")

# Paso 2: Edad
while True:
    try:
        edad = int(input(f"👋 ¡Genial, {nombre}! ¿Cuántos años tienes? "))
        if 5 <= edad <= 100:
            break
        print("⚠️ Ingresa una edad entre 5 y 100 años.")
    except ValueError:
        print("⚠️ Por favor ingresa solo un número.")

if edad < 13:
    mensaje_edad = "¡Eres un/a explorador/a junior perfecto/a! 🎮"
    distro = "Ubuntu (fácil y divertido)"
elif edad < 18:
    mensaje_edad = "¡Estás en la edad ideal para dominar Linux! 🚀"
    distro = "Linux Mint (moderno y amigable)"
else:
    mensaje_edad = "¡Nunca es tarde para aprender! 🌟"
    distro = "Debian (estable y poderoso)"

print(f"\n{mensaje_edad}")
print(f"💻 Te recomiendo empezar con: {distro}")

# Paso 3: Experiencia
print("\n🤔 ¿Qué sabes de Linux?")
print("1. 🐧 Ya conozco a Tux y he usado Linux")
print("2. 🤔 He oído hablar, pero no lo he probado")
print("3. 🆕 Es totalmente nuevo para mí")

while True:
    opcion = input("\nElige 1, 2 o 3: ").strip()
    if opcion == "1":
        experiencia = "Explorador/a Avanzado/a 🔥"
        break
    elif opcion == "2":
        experiencia = "Aprendiz Curioso/a ✨"
        break
    elif opcion == "3":
        experiencia = "Nuevo/a Aventurero/a 🎯"
        break
    print("⚠️ Por favor elige solo 1, 2 o 3.")

# Paso 4: Lugar favorito
lugares = [
    "Volcán de Izalco 🌋",
    "Playa El Tunco 🏖️",
    "Lago de Coatepeque 🏞️",
    "San Salvador 🏛️",
    "Ruta de las Flores 🌸",
    "Suchitoto 🏰",
    "Playa El Zonte 🌊",
    "Joya de Cerén 🗿",
    "Otro lugar 📍"
]

print("\n🏞️ Tu lugar favorito en El Salvador:")
for i, lugar in enumerate(lugares, 1):
    print(f"{i}. {lugar}")

while True:
    try:
        sel = int(input("\nElige un número (1-9): "))
        if 1 <= sel <= 9:
            if sel == 9:
                lugar_favorito = input("\n✏️ Escribe tu lugar favorito: ").strip()
                if lugar_favorito:
                    break
            else:
                lugar_favorito = lugares[sel-1].split(" ", 1)[0]
                break
    except ValueError:
        pass
    print("⚠️ Ingresa un número válido del 1 al 9.")

print(f"\n¡Qué lindo lugar: {lugar_favorito}! Tux quiere ir de vacaciones ahí. 🏝️")

# Resumen final
print("\n" + "=" * 80)
print("🎉 ¡TU PERFIL DE AVENTURERO/A ESTÁ LISTO! ".center(80))
print("=" * 80)
print(f"┌{'─'*70}┐")
print(f"│ 👤 Nombre: {nombre:<55} │")
print(f"│ 🎂 Edad: {edad} años{'':<48} │")
print(f"│ 🏆 Nivel Linux: {experiencia:<47} │")
print(f"│ 🇸🇻 Lugar favorito: {lugar_favorito:<44} │")
print(f"│ 💻 Distro recomendada: {distro:<39} │")
print(f"└{'─'*70}┘\n")

print("💡 DATOS CURIOSOS SOBRE LINUX:")
for dato in [
    "• El 100% de las supercomputadoras más rápidas usan Linux",
    "• Android (tu celular) está basado en Linux",
    "• La NASA usa Linux en el espacio 🚀",
    "• Linux es GRATIS y libre para siempre ❤️"
]:
    print(dato)

print("\n🎯 TU PRIMERA MISIÓN:")
print("1. Abre una terminal")
print(f"2. Crea una carpeta: mkdir Aventura_Linux_{nombre.split()[0].lower()}")
print("3. Entra a la carpeta y crea un archivo con tu perfil usando nano o touch")
print("\n¡Tux te está esperando! 🐧")
print(f"\n🎊 ¡Felicidades, {nombre}! Tu aventura con Linux ha comenzado. 🇸🇻💚")
````
