### Paso 2: Escribe el script mágico

Ejecuta esta celda para escribir todo el script de una vez. ¡Es tu obra maestra!

```python
%%bash
cat > aventura_tux.sh << 'EOF'
#!/bin/bash

# Colores mágicos
VERDE='\033[0;32m'
AZUL='\033[0;34m'
AMARILLO='\033[1;33m'
ROJO='\033[0;31m'
NORMAL='\033[0m'

clear
echo "${AZUL}=======================================${NORMAL}"
echo "${VERDE}   ¡Bienvenido a la Aventura de Tux!   ${NORMAL}"
echo "${AZUL}=======================================${NORMAL}"
echo "${AMARILLO}Tux te está esperando en El Salvador... 🐧🇸🇻${NORMAL}"
sleep 2

echo "\n${VERDE}¿Cómo te llamas, valiente aventurero/a?${NORMAL}"
read nombre

echo "\n${VERDE}¡Hola $nombre! Tux está muy emocionado de verte.${NORMAL}"
echo "${AMARILLO}Estás cerca de tu lugar favorito: ${NORMAL}"
read lugar

echo "\n${AZUL}¡Genial! Tux ha construido su casita cerca de $lugar.${NORMAL}"
sleep 2

echo "\n${VERDE}Tux te pregunta: ¿Qué quieres hacer hoy?${NORMAL}"
echo "1) Explorar la casa de Tux"
echo "2) Buscar un tesoro secreto"
echo "3) Despedirse y terminar la aventura"
read opcion

if [ "$opcion" = "1" ]; then
    echo "\n${AZUL}¡Entramos a la casa de Tux! 🏠${NORMAL}"
    echo "Ves habitaciones llenas de comandos mágicos... ¡Qué divertido!"
elif [ "$opcion" = "2" ]; then
    echo "\n${AMARILLO}¡Buscando tesoro...! 🔍${NORMAL}"
    sleep 2
    echo "${VERDE}¡LO ENCONTRASTE! Es una foto de pupusas y $lugar 🌋🍽️${NORMAL}"
else
    echo "\n${ROJO}¡Adiós $nombre! Tux te espera para la próxima aventura 🐧❤️${NORMAL}"
    exit
fi

echo "\n${VERDE}¡Gracias por jugar con Tux, $nombre!${NORMAL}"
echo "${AZUL}Has completado la gran aventura de Linux en El Salvador 🇸🇻${NORMAL}"
echo "${AMARILLO}¡Eres un verdadero maestro/a de Linux ahora! 🎉🐧${NORMAL}"
EOF
```

### Paso 3: Dale permisos de ejecución

Para que el script se pueda correr como un programa.

```python
!chmod +x aventura_tux.sh
```

### Paso 4: ¡Ejecuta tu aventura!

```python
!./aventura_tux.sh
```

```python
# Importaciones necesarias
import ipywidgets as widgets
from datetime import datetime
from IPython.display import display, HTML, clear_output
import base64
import io

# === INTENTAMOS IMPORTAR WEASYPRINT (MEJOR OPCIÓN PARA PDF) ===
try:
    from weasyprint import HTML as WPHTML
    WEASYPRINT_AVAILABLE = True
except ImportError:
    WEASYPRINT_AVAILABLE = False
    print("⚠️ Para PDF de máxima calidad: !pip install weasyprint")

# --- WIDGETS (igual que antes, pero sin botón de PNG) ---
nombre_certificado = widgets.Text(placeholder='Ej: Maria Guadalupe Hernández', description='Nombre completo:', style={'description_width': 'initial'})
centro_escolar = widgets.Text(placeholder='Ej: Centro Escolar República de Guatemala', description='Centro escolar:', style={'description_width': 'initial'})
grado_selector = widgets.Dropdown(options=[('Primer grado', '1° Grado'), ('Segundo grado', '2° Grado'), ('Tercer grado', '3° Grado'), ('Cuarto grado', '4° Grado'), ('Quinto grado', '5° Grado'), ('Sexto grado', '6° Grado'), ('Séptimo grado', '7° Grado'), ('Octavo grado', '8° Grado'), ('Noveno grado', '9° Grado'), ('Bachillerato', 'Bachillerato')], value='6° Grado', description='Grado:', style={'description_width': 'initial'})
ciudad_certificado = widgets.Text(placeholder='Ej: San Salvador, Santa Ana...', description='Ciudad:', style={'description_width': 'initial'})
edad_certificado = widgets.IntSlider(value=12, min=5, max=20, description='Edad:', style={'description_width': 'initial'})

boton_generar = widgets.Button(description='GENERAR CERTIFICADO', button_style='success', icon='graduation-cap', layout=widgets.Layout(width='220px'))
boton_descargar_pdf = widgets.Button(description='DESCARGAR PDF', button_style='warning', icon='file-pdf', layout=widgets.Layout(width='200px'), disabled=True)

output_certificado = widgets.Output()

# Variable global para guardar el HTML del certificado
certificado_html_content = ""

def generar_certificado(b):
    global certificado_html_content
    with output_certificado:
        clear_output()
        
        if not all([nombre_certificado.value.strip(), centro_escolar.value.strip(), ciudad_certificado.value.strip()]):
            print("❌ Por favor, completa todos los campos obligatorios")
            return
        
        nombre = nombre_certificado.value.strip()
        centro = centro_escolar.value.strip()
        grado = grado_selector.value
        ciudad = ciudad_certificado.value.strip()
        edad = edad_certificado.value
        fecha = datetime.now().strftime("%d de %B de %Y")
        
        boton_descargar_pdf.disabled = False
        
        # === HTML DEL CERTIFICADO (ajustado para que quepa en una página) ===
        certificado_html_content = f"""
        <div style="background: linear-gradient(135deg, #2C3E50, #3498db); padding: 40px; border-radius: 20px; border: 15px solid #FFD700; box-shadow: 0 15px 30px rgba(0,0,0,0.3); font-family: 'Segoe UI', sans-serif; max-width: 700px; margin: 0 auto; overflow: hidden;">
            <!-- Todo tu diseño hermoso aquí (ajustado para tamaño carta) -->
            <div style="border: 8px solid #FFD700; padding: 15px;">
                <div style="border: 8px solid #FFD700; padding: 15px;">
                    <div style="border: 8px solid #FFD700; padding: 15px;">
                        <div style="background: radial-gradient(circle, #FFD700, #B8860B); width: 90px; height: 90px; border-radius: 50%; margin: -30px auto 15px; display: flex; align-items: center; justify-content: center; box-shadow: 0 5px 15px rgba(0,0,0,0.3);">
                            <span style="font-size: 45px; color: white;">🏆</span>
                        </div>
                        <h1 style="text-align: center; color: white; font-size: 28px; margin: 8px 0;">CERTIFICADO DE LOGRO</h1>
                        <p style="text-align: center; color: white; font-size: 16px;">Reconocimiento Académico y Tecnológico</p>
                        <div style="background: #90EE90; padding: 30px; border-radius: 15px; margin: 20px 0; border: 6px solid #90EE90;">
                            <h2 style="text-align: center; color: #2C3E50; font-size: 22px;">OFICIAL</h2>
                            <p style="text-align: center; color: #2C3E50; font-size: 20px; font-weight: bold;">{nombre}</p>
                            <p style="text-align: center; color: #2C3E50; font-size: 16px;">Por completar exitosamente el</p>
                            <h3 style="text-align: center; color: #2C3E50; font-size: 20px;">MÓDULO 5: Programar Jugando con Python 🐍</h3>
                            <div style="text-align: center; margin: 20px 0; line-height: 1.6;">
                                <p style="font-size: 14px;">🎓 Centro Escolar: {centro}</p>
                                <p style="font-size: 14px;">📍 {ciudad}</p>
                                <p style="font-size: 14px;">📚 Grado: {grado}</p>
                                <p style="font-size: 14px;">🎂 Edad: {edad} años</p>
                                <p style="font-size: 14px;">📅 Fecha: {fecha}</p>
                            </div>
                            <div style="background: linear-gradient(90deg, #2C3E50, #3498db); padding: 20px; border-radius: 10px; border: 3px solid #5D3FD3;">
                                <h4 style="text-align: center; color: white;">HABILIDADES DEMOSTRADAS</h4>
                                <div style="display: grid; grid-template-columns: 1fr 1fr; gap: 10px; margin-top: 10px;">
                                    <!-- Usamos fondo blanco sólido para mejor renderizado -->
                                    <div style="background: white; padding: 8px; border-radius: 8px; text-align: center; color: #2C3E50; font-weight: bold; font-size: 13px;">✅ Uso de print()</div>
                                    <div style="background: white; padding: 8px; border-radius: 8px; text-align: center; color: #2C3E50; font-weight: bold; font-size: 13px;">✅ Variables en Python</div>
                                    <div style="background: white; padding: 8px; border-radius: 8px; text-align: center; color: #2C3E50; font-weight: bold; font-size: 13px;">✅ Bucles con for</div>
                                    <div style="background: white; padding: 8px; border-radius: 8px; text-align: center; color: #2C3E50; font-weight: bold; font-size: 13px;">✅ Resolución de problemas</div>
                                </div>
                            </div>
                            <p style="text-align: center; color: #2C3E50; font-style: italic; margin-top: 20px; font-size: 14px;">
                                "El futuro pertenece a quienes creen en la belleza de sus sueños y tienen el valor de programarlos!"
                            </p>
                        </div>
                        <div style="text-align: center; color: white; margin-top: 20px;">
                            <p style="font-size: 14px;">👨‍💻 Aprende Linux desde Cero</p>
                            <p style="font-size: 12px;">Programa educativo para El Salvador</p>
                            <p style="font-size: 12px;">wolfsv.uncurled353@aleeas.com</p>
                            <p style="font-size: 12px;">🌐 https://satscoinsv.bearblog.dev/</p>
                        </div>
                    </div>
                </div>
            </div>
        </div>
        """
        
        full_html = f"""
        <html><head><meta charset="utf-8"></head><body style="margin:0; padding:10px; background:#f0f0f0;">{certificado_html_content}</body></html>
        """
        
        display(HTML(full_html))
        print("✅ ¡Certificado generado! Ahora puedes descargarlo en PDF.")

def descargar_pdf(b):
    with output_certificado:
        clear_output()
        
        if not certificado_html_content:
            print("❌ Primero genera el certificado")
            return
        
        if WEASYPRINT_AVAILABLE:
            print("📄 Generando PDF de alta calidad...")
            
            # HTML optimizado para impresión en tamaño carta con márgenes ajustados
            html_pdf = f"""
            <html>
            <head>
                <meta charset="utf-8">
                <style>
                    @page {{ 
                        size: letter; 
                        margin: 1cm; 
                        bleed: 5mm;
                    }}
                    body {{ 
                        margin: 0; 
                        padding: 0;
                        background: #f0f0f0;
                    }}
                    .certificate {{ 
                        width: 100%; 
                        max-width: 600px; 
                        margin: 0 auto;
                        background: linear-gradient(135deg, #2C3E50, #3498db);
                        padding: 30px;
                        border-radius: 15px;
                        border: 12px solid #FFD700;
                        box-shadow: 0 10px 20px rgba(0,0,0,0.3);
                        overflow: hidden;
                    }}
                    .inner-border {{ 
                        background: white; 
                        padding: 25px; 
                        border-radius: 10px;
                        border: 6px solid #90EE90;
                    }}
                    .skills-grid {{ 
                        display: grid; 
                        grid-template-columns: 1fr 1fr; 
                        gap: 8px; 
                        margin-top: 12px;
                    }}
                    .skill-box {{ 
                        background: white; 
                        padding: 8px; 
                        border-radius: 6px; 
                        text-align: center; 
                        color: #2C3E50; 
                        font-weight: bold;
                        font-size: 12px;
                    }}
                    h1 {{ font-size: 24px; }}
                    h2 {{ font-size: 20px; }}
                    h3 {{ font-size: 18px; }}
                    h4 {{ font-size: 16px; }}
                    p {{ font-size: 12px; }}
                </style>
            </head>
            <body>
                <div class="certificate">
                    <div style="text-align: center; margin-bottom: 15px;">
                        <div style="background: radial-gradient(circle, #FFD700, #B8860B); width: 80px; height: 80px; border-radius: 50%; margin: 0 auto 12px; display: flex; align-items: center; justify-content: center; box-shadow: 0 3px 10px rgba(0,0,0,0.3);">
                            <span style="font-size: 40px; color: white;">🏆</span>
                        </div>
                        <h1 style="color: white;">CERTIFICADO DE LOGRO</h1>
                        <p style="color: white; font-size: 14px;">Reconocimiento Académico y Tecnológico</p>
                    </div>
                    
                    <div class="inner-border">
                        <h2 style="text-align: center; color: #2C3E50;">OFICIAL</h2>
                        <p style="text-align: center; color: #2C3E50; font-size: 18px; font-weight: bold;">{nombre_certificado.value.strip()}</p>
                        <p style="text-align: center; color: #2C3E50; font-size: 14px;">Por completar exitosamente el</p>
                        <h3 style="text-align: center; color: #2C3E50;">MÓDULO 5: Programar Jugando con Python 🐍</h3>
                        
                        <div style="text-align: center; margin: 15px 0; line-height: 1.4;">
                            <p style="font-size: 12px;">🎓 Centro Escolar: {centro_escolar.value.strip()}</p>
                            <p style="font-size: 12px;">📍 {ciudad_certificado.value.strip()}</p>
                            <p style="font-size: 12px;">📚 Grado: {grado_selector.value}</p>
                            <p style="font-size: 12px;">🎂 Edad: {edad_certificado.value} años</p>
                            <p style="font-size: 12px;">📅 Fecha: {datetime.now().strftime("%d de %B de %Y")}</p>
                        </div>
                        
                        <div style="background: linear-gradient(90deg, #2C3E50, #3498db); padding: 15px; border-radius: 8px; border: 2px solid #5D3FD3;">
                            <h4 style="text-align: center; color: white;">HABILIDADES DEMOSTRADAS</h4>
                            <div class="skills-grid">
                                <div class="skill-box">✅ Uso de print()</div>
                                <div class="skill-box">✅ Variables en Python</div>
                                <div class="skill-box">✅ Bucles con for</div>
                                <div class="skill-box">✅ Resolución de problemas</div>
                            </div>
                        </div>
                        
                        <p style="text-align: center; color: #2C3E50; font-style: italic; margin-top: 15px; font-size: 12px;">
                            "El futuro pertenece a quienes creen en la belleza de sus sueños y tienen el valor de programarlos!"
                        </p>
                    </div>
                    
                    <div style="text-align: center; color: white; margin-top: 15px;">
                        <p style="font-size: 12px;">👨‍💻 Aprende Linux desde Cero</p>
                        <p style="font-size: 10px;">Programa educativo para El Salvador</p>
                        <p style="font-size: 10px;">wolfsv.uncurled353@aleeas.com</p>
                        <p style="font-size: 10px;">🌐 https://satscoinsv.bearblog.dev/</p>
                    </div>
                </div>
            </body>
            </html>
            """
            
            pdf_io = io.BytesIO()
            WPHTML(string=html_pdf).write_pdf(pdf_io)
            pdf_io.seek(0)
            b64 = base64.b64encode(pdf_io.read()).decode()
            
            filename = f"Certificado_{nombre_certificado.value.replace(' ', '_')}_{datetime.now().strftime('%Y%m%d')}.pdf"
            
            display(HTML(f"""
            <div style="text-align:center; padding:30px; background:#d4edda; border-radius:10px;">
                <h3>✅ ¡PDF listo!</h3>
                <a href="data:application/pdf;base64,{b64}" download="{filename}"
                   style="background:#28a745; color:white; padding:15px 30px; text-decoration:none; border-radius:8px; font-size:18px;">
                    📥 DESCARGAR CERTIFICADO PDF
                </a>
            </div>
            """))
        else:
            print("❌ WeasyPrint no disponible. Por favor instala con: pip install weasyprint")

# Conectar botones
boton_generar.on_click(generar_certificado)
boton_descargar_pdf.on_click(descargar_pdf)

# Formulario final
display(widgets.VBox([
    widgets.HTML("<h2 style='text-align:center; color:#2C3E50;'>🎓 GENERADOR DE CERTIFICADOS</h2>"),
    nombre_certificado, centro_escolar, grado_selector, ciudad_certificado, edad_certificado,
    widgets.HTML("<hr>"),
    widgets.HBox([boton_generar, boton_descargar_pdf], layout=widgets.Layout(justify_content='center')),
    output_certificado
]))
```
