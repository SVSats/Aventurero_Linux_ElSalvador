**Explicación de los candados mágicos:**

La primera columna muestra algo como: `-rw-r--r--`

- El primer carácter: `-` = archivo normal, `d` = carpeta
- Luego 3 grupos de 3 letras:
  1. **Dueño** (tú): r = leer, w = escribir, x = ejecutar
  2. **Grupo** (amigos)
  3. **Otros** (el resto del mundo)

`r` = leer (abrir y ver)  
`w` = escribir (modificar)  
`x` = ejecutar (correr si es un programa)  
`-` = sin permiso (candado cerrado)

Normalmente tus archivos nuevos tienen: `rw-r--r--` → tú puedes todo, los demás solo leer.

### Comando principal: `chmod` → ¡Cambia los candados mágicos!

`chmod` = **Change Mode** (cambiar permisos).

Hay dos formas fáciles: con números o con letras.

#### Forma con números (la más fácil de recordar)

- 4 = leer (r)
- 2 = escribir (w)
- 1 = ejecutar (x)
- 0 = nada

Se suman: leer + escribir = 6, todo = 7

Ejemplos comunes:
- 644 → dueño todo menos ejecutar, otros solo leer
- 755 → dueño todo, otros leer y ejecutar (para carpetas y programas)
- 600 → solo dueño lee y escribe (¡super secreto!)

```python
# Haz el tesoro_secreto súper privado (solo tú lees y escribes)
!chmod 600 tesoro_secreto.txt
```

```python
!ls -l
```

```python
# ¡Ahora haz la receta pública para compartir con todos!
!chmod 644 receta_pupusas.txt
```

#### Forma con letras (más clara para entender)

- u = usuario (dueño)
- g = grupo
- o = otros
- a = todos

+ para agregar, - para quitar

```python
# Agrega permiso de ejecución al mapa (como si fuera un programa)
!chmod u+x mapa_volcan.txt
```

```python
!ls -l
```
