### Comando #1: `ip a` o `ifconfig` → ¡Mira tu tarjeta de red (tu pasaporte a internet)!

Este comando muestra tus direcciones IP (como la dirección de tu casa en internet).

```python
# Comando moderno (recomendado)
!ip a
```

```python
# Comando clásico (si tienes ifconfig instalado)
# !ifconfig
```

Busca la interfaz que dice `wlan0` (WiFi) o `eth0` (cable). Ahí está tu dirección IP local (empieza con 192.168... o 10...).

¡Esa es la dirección de tu computadora en la red local de tu casa!

### Comando #2: `ping` → ¡Envía saludos a un amigo y ve si responde!

`ping` envía paquetes pequeños para ver si un sitio está conectado y cuánto tarda en responder.

¡Vamos a saludar a Google desde El Salvador!

```python
# Envía 6 saludos a Google (presiona Ctrl+C para detener antes)
!ping -c 6 google.com
```

```python
# Prueba con un sitio salvadoreño
!ping -c 4 www.elsalvador.com
```

Mira el tiempo en `time=... ms`. ¡Cuánto más bajo, más rápido!

Si dice "Destination Host Unreachable" → no hay conexión o el sitio no existe.

### Comando #3: `curl` → ¡Descarga páginas web directamente!

`curl` trae el contenido de una página web a la terminal.

¡Vamos a ver la página principal de Wikipedia en español!

```python
# Muestra solo los primeros líneas de Wikipedia
!curl -s https://es.wikipedia.org/wiki/El_Salvador | head -20
```

¡Es HTML! El código que forma la página web.

También puedes guardar una página completa:

```python
# Guarda la página de Tux en Linux
!curl -o pagina_tux.html https://es.wikipedia.org/wiki/Tux
```

```python
# ¡Mira que se guardó!
!ls -l pagina_tux.html
```

### Comando #4: `wget` → ¡Descarga archivos completos!

`wget` descarga archivos (imágenes, documentos, etc.).

¡Vamos a descargar una foto de Tux!

```python
# Descarga una imagen pequeña de Tux
!wget -O mi_tux.png https://upload.wikimedia.org/wikipedia/commons/thumb/3/35/Tux.svg/200px-Tux.svg.png
```

```python
!ls -l mi_tux.png
```
