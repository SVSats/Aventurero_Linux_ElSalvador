### Primero: Entra a tu carpeta secreta

Recuerda la carpeta que creaste en el módulo anterior: **Tesoros_ElSalvador**  
¡Vamos a entrar ahí para guardar nuestros mensajes secretos!

````python
# Entra a la carpeta secreta
!cd Tesoros_ElSalvador 2>/dev/null || mkdir Tesoros_ElSalvador && cd Tesoros_ElSalvador
!pwd
!ls
````

### Comando #1: `touch` → ¡Crea un archivo vacío como por arte de magia!

`touch` crea un archivo nuevo (como una hoja en blanco).

¡Vamos a crear una carta secreta para Tux!

````python
# Crea un archivo llamado carta_a_tux.txt
!touch carta_a_tux.txt
````

````python
# ¡Mira! Ya aparece el archivo
!ls
````

### Comando #2: `echo` → ¡Escribe un mensaje rápido en el archivo!

`echo` imprime texto, y con `>` lo guarda en un archivo.

¡Vamos a escribir un saludo salvadoreño para Tux!

````python
# Escribe un mensaje en la carta
!echo "¡Hola Tux! Soy {nombre} de El Salvador y me encanta Linux 🐧" > carta_a_tux.txt
````

### Comando #3: `cat` → ¡Lee el mensaje secreto!

`cat` significa **concatenate** (mostrar el contenido).

¡Es como abrir la carta y leerla!

````python
# Lee la carta
!cat carta_a_tux.txt
````

¡Guau! Ahí está tu mensaje personalizado. 😄

### Agreguemos más líneas con `>>`

`>>` agrega texto al final del archivo (sin borrarlo).

````python
!echo "Mi lugar favorito es {lugar_favorito}" >> carta_a_tux.txt
!echo "¡Quiero aprender más comandos mágicos!" >> carta_a_tux.txt
!echo "Atentamente, {nombre} ({edad} años)" >> carta_a_tux.txt
````

````python
# ¡Lee la carta completa ahora!
!cat carta_a_tux.txt
````

### Comando #4: `nano` → ¡Edita el archivo como un profesional!

`nano` es un editor de texto súper fácil dentro de la terminal.

**Nota:** En Jupyter el comando `nano` no funciona directamente (es interactivo), pero te muestro cómo usarlo en la terminal real.

Abre una terminal normal y escribe:

```bash
cd Tesoros_ElSalvador
nano carta_a_tux.txt
```

Dentro de nano:
- Escribe lo que quieras
- Para guardar: Ctrl + O → Enter
- Para salir: Ctrl + X

- `touch` → crear archivos
- `echo >` → escribir en un archivo
- `echo >>` → agregar más texto
- `cat` → leer archivos
- `nano` → editar como un pro
