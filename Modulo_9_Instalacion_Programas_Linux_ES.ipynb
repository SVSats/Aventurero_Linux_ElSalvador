### Primero: Actualiza la lista de programas disponibles

Antes de instalar nada, es buena idea actualizar la "tienda" de programas.

**Comando mágico:** `sudo apt update`

**¡Importante!** El comando `sudo` significa "superusuario do" (hacer como administrador). Te pedirá tu contraseña.

En Jupyter no podemos usar `sudo` fácilmente, pero te muestro cómo se hace en la terminal real:

```bash
# En una terminal real:
sudo apt update          # Actualiza la lista de programas
sudo apt upgrade         # Instala las actualizaciones disponibles (opcional)
```

### Comando #1: `apt` → ¡La tienda oficial de Ubuntu!

`apt` es el administrador de paquetes de Ubuntu/Debian. Tiene miles de programas gratis.

Ejemplos divertidos (ejecútalos en una terminal real con `sudo`):

```bash
# Busca programas
apt search minecraft
apt search gimp          # Editor de fotos como Photoshop
apt search vlc           # Reproductor de videos

# Instala un programa
sudo apt install vlc     # Instala VLC
sudo apt install gimp    # Instala GIMP
sudo apt install scratch # Instala Scratch para programar juegos

# Desinstala si ya no lo quieres
sudo apt remove vlc
```

### Comando #2: `snap` → ¡Programas modernos en paquetes mágicos!

`snap` instala programas en "contenedores" que siempre funcionan, sin importar la versión de Ubuntu.

¡Muchos programas famosos vienen en snap!

```bash
# Busca snaps
snap find minecraft
snap find spotify
snap find code           # Visual Studio Code

# Instala un snap
sudo snap install spotify
sudo snap install code --classic
sudo snap install minecraft-launcher-official

# Lista tus snaps instalados
snap list
```

### Comando #3: `flatpak` → ¡Otra tienda genial! (opcional)

Si tienes Flatpak instalado, puedes usarlo también (muchos juegos y programas).

```bash
# Instala Flatpak primero (una vez)
sudo apt install flatpak

# Agrega la tienda Flathub
flatpak remote-add --if-not-exists flathub https://flathub.org/repo/flathub.flatpakrepo

# Instala programas
flatpak install flathub org.minecraft.Minecraft
```
