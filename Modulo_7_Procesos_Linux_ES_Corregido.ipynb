### Comando #1: `ps` → ¡Mira los procesos como una foto instantánea!

`ps` = **Process Status** (estado de procesos).

¡Es como tomar una foto de todo lo que está haciendo Tux ahora mismo!

```python
# Muestra todos los procesos del sistema (puede ser largo)
!ps aux | head -20
```

```python
# Solo los procesos de tu usuario (más limpio)
!ps aux | grep $USER | grep -v grep
```

Mira las columnas importantes:
- **PID**: número único de cada proceso (como su cédula)
- **%CPU**: cuánto procesador usa
- **%MEM**: cuánta memoria usa
- **COMMAND**: nombre del programa

### Comando #2: `top` → ¡Ve los procesos en vivo!

`top` muestra los procesos actualizados constantemente.

En Jupyter no es interactivo, pero podemos ver unas cuantas actualizaciones.

```python
# Muestra top 3 veces (como una pequeña película)
!top -b -n 3 -d 2
```

¡Mira cómo cambian los números! Los procesos que más CPU usan aparecen arriba.

**Tip:** En una terminal real, escribe `top` y presiona `q` para salir. Si tienes `htop` instalado, ¡es aún más bonito y colorido!

### Comando #3: `kill` → ¡Detén un proceso rebelde! (explicación segura)

`kill` envía una señal a un proceso para que se detenga.

**¡Importante!** Solo úsalo cuando un programa se congela y no responde.

Como en Jupyter no podemos crear procesos en segundo plano fácilmente, te explico cómo usarlo en la vida real:

1. Abre una terminal normal.
2. Ejecuta: `sleep 300 &` (crea un proceso dormilón)
3. En otra terminal: `ps aux | grep sleep` → busca el PID
4. Detenlo con: `kill PID` (amable) o `kill -9 PID` (fuerte, si no obedece)

**Ejemplo seguro aquí en Jupyter (sin background):**

```python
# Esto es solo una demostración (no crea un proceso detachable)
print("Imagina que encontramos un proceso con PID 12345 que se congeló")
print("Lo detendríamos con: kill 9 12345")
print("¡Pero NUNCA lo uses sin saber el PID exacto!")
```

**Reglas de oro del jefe:**
- Usa `ps` o `top` para encontrar el PID correcto
- Prueba primero `kill PID` (señal amable)
- Solo usa `kill -9 PID` si no responde (es como apagar de emergencia)
- ¡Nunca mates procesos del sistema si no sabes qué haces!
