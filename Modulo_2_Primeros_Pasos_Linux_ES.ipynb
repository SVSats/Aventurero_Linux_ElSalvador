### Comando #1: `pwd` → ¿Dónde estoy parado?

`pwd` significa **Print Working Directory** (imprime dónde estoy).

¡Es como preguntarle a Tux: "¿En qué habitación estoy ahora mismo?" 🧭

```python
!pwd
```

¡Mira la ruta que aparece! Esa es tu ubicación actual.  
Normalmente empiezas en tu carpeta personal: `/home/tu_usuario` (es como tu cuarto en la gran casa de Linux).

### Comando #2: `ls` → ¿Qué hay en esta habitación?

`ls` significa **List** (lista lo que hay).

¡Es como encender la luz y ver todos los muebles, puertas y juguetes! 👀

```python
!ls
```

Las carpetas suelen aparecer en **azul**, los archivos en blanco o colores.  
¡Prueba esta versión más poderosa para ver más detalles!

```python
!ls -la
```

### Comando #3: `cd` → ¡Muévete a otra habitación!

`cd` significa **Change Directory** (cambiar de carpeta).

Ejemplos mágicos:
- `cd Escritorio` → entra al escritorio
- `cd ..` → sube un nivel (como salir al pasillo)
- `cd ~` → vuelve a tu carpeta personal
- `cd /` → va al inicio de toda la casa Linux

```python
# Prueba moverte al Escritorio (si existe) y ver qué hay
!cd Escritorio 2>/dev/null || echo "No existe la carpeta Escritorio, prueba con otra"
!pwd
!ls
```

¡Vamos a crear una habitación secreta para guardar tesoros salvadoreños!
