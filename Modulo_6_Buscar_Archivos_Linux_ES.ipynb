### Primero: Preparamos el terreno de búsqueda

Vamos a crear algunas carpetas y archivos de práctica para que puedas buscar como un pro.

```python
# Creamos una zona de búsqueda con carpetas y tesoros escondidos
!mkdir -p Busqueda_Tesoros/volcanes playa jardin secreto
!echo "Mapa del Volcán de Izalco" > Busqueda_Tesoros/volcanes/mapa_izalco.txt
!echo "Foto de la playa El Tunco" > Busqueda_Tesoros/playa/foto_tunco.txt
!echo "Receta secreta de pupusas" > Busqueda_Tesoros/secreto/receta_pupusas.txt
!echo "Pupusas deliciosas con curtido" >> Busqueda_Tesoros/secreto/receta_pupusas.txt
!echo "Foto del Lago de Coatepeque" > Busqueda_Tesoros/jardin/foto_coatepeque.jpg
!touch Busqueda_Tesoros/volcanes/notas_antiguas.txt
```

```python
# ¡Mira cómo quedó el terreno!
!ls -R Busqueda_Tesoros
```

### Comando #1: `find` → ¡Encuentra archivos perdidos por nombre o tipo!

`find` es como un perro rastreador que busca en todas las carpetas.

¡Vamos a buscar todos los archivos que hablen de volcanes!

```python
# Busca archivos por nombre
!find Busqueda_Tesoros -name "*volc*"
```

```python
# Busca solo archivos .txt
!find Busqueda_Tesoros -type f -name "*.txt"
```

```python
# Busca solo carpetas
!find Busqueda_Tesoros -type d
```

### Comando #2: `grep` → ¡Busca palabras secretas dentro de los archivos!

`grep` busca texto dentro de archivos. ¡Es como un detector de palabras mágicas!

¡Vamos a buscar dónde está escondida la palabra "pupusas"!

```python
# Busca "pupusas" en todos los archivos de la zona
!grep -r "pupusas" Busqueda_Tesoros/
```

```python
# Busca mostrando el número de línea
!grep -n "pupusas" Busqueda_Tesoros/secreto/receta_pupusas.txt
```

```python
# Busca ignorando mayúsculas/minúsculas
!grep -i "COATEPEQUE" Busqueda_Tesoros/
```

### Comando #3: `wc` → ¡Cuenta líneas, palabras y caracteres!

`wc` = **Word Count** (contar palabras).

¡Perfecto para saber cuántos secretos tiene un archivo!

```python
# Cuenta todo en la receta secreta
!wc Busqueda_Tesoros/secreto/receta_pupusas.txt
```

```python
# Solo líneas (-l), palabras (-w) o caracteres (-c)
!wc -l Busqueda_Tesoros/secreto/receta_pupusas.txt
!wc -w Busqueda_Tesoros/secreto/receta_pupusas.txt
```

### ¡Super combinación: find + grep!

¡Los detectives avanzados usan varios comandos juntos!

```python
# Encuentra todos los .txt y busca dentro la palabra "foto"
!find Busqueda_Tesoros -name "*.txt" -exec grep -l "foto" {} \;
```
